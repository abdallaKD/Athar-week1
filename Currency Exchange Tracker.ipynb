```python
import os
import sys
from pathlib import Path
from dotenv import load_dotenv
import requests
```

```python
print("Python:", sys.executable)
print("Project folder:", Path.cwd())

load_dotenv(r"D:\Athar\Week1\.env")

api_key = os.getenv("EXCHANGE_RATE_API_KEY")

if api_key:
    print("API Key loaded successfully!")
else:
    print("ERROR: API Key not found!")
```

```python
url = "https://v6.exchangerate-api.com/v6/latest/USD"

headers = {
    "Authorization": f"Bearer {api_key}"
}

# Send GET request
response = requests.get(url, headers=headers, timeout=15)

# Check HTTP status
response.raise_for_status()

# Convert JSON response to Python dictionary
data = response.json()

# Check API result
if data.get("result") == "success":
    print("API Connection Successful!")
    print("----------------------------")
    print("Base Currency:", data["base_code"])
    print("Last Update:", data["time_last_update_utc"])
    print("USD to EGP:", data["conversion_rates"]["EGP"])
    print("USD to EUR:", data["conversion_rates"]["EUR"])
    print("USD to GBP:", data["conversion_rates"]["GBP"])
else:
    print("API Error:", data.get("error-type"))
```

```python

class ExchangeRates:

    def __init__(self, base_currency, last_update, rates):
        self.base_currency = base_currency
        self.last_update = last_update
        self.rates = rates

    def get_rate(self, currency):
        return self.rates[currency.upper()]

    def print_summary(self):
        print("=" * 45)
        print("       CURRENCY EXCHANGE RATE TRACKER")
        print("=" * 45)

        print(f"Base Currency: {self.base_currency}")
        print(f"Last Update: {self.last_update}")
        print("-" * 45)

        for currency in ["EGP", "EUR", "GBP"]:
            rate = self.get_rate(currency)
            print(f"1 {self.base_currency} = {rate:.4f} {currency}")

        print("-" * 45)

        amount = 100
        converted = amount * self.get_rate("EGP")

        print(f"{amount} USD = {converted:,.2f} EGP")
        print("=" * 45)


# Create an object from the API JSON data
exchange = ExchangeRates(
    base_currency=data["base_code"],
    last_update=data["time_last_update_utc"],
    rates=data["conversion_rates"]
)

# Display formatted summary
exchange.print_summary()
```
